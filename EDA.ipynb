# Análisis Exploratorio de Datos (EDA)

Exploraremos los datos limpios para comprender las relaciones entre las variables de supervivenia.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('data/titanic_clean.csv')
df.head()
```

```python
df.sample(4)
```

# Supervivencia por Género

```python
df['Survived'].value_counts()
```

```python
plt.figure(figsize=(12, 8))

sns.barplot(x = df['Survived'].value_counts().index, y = df['Survived'].value_counts())
plt.title('Value Counts of Survived')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Sex', y='Survived', data=df, palette='viridis')
plt.title('Probabilidad de Supervivencia por Género')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Pclass', y= 'Survived', data=df, palette='magma')
plt.title('Probabilidad de Supervivencia por Clase')
plt.show()
```

```python
plt.figure(figsize = (12, 8))

# colores para cada una de las clases
color = ['#240750', '#344C64']

sns.countplot(df, x = 'Embarked', hue = 'Survived', palette=color) 
plt.legend(title = 'Supervivencia', labels = ['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

# Distribución de Edades

```python
plt.figure(figsize = (12, 8))

sns.histplot(df, x = 'Age', kde = True)
plt.title('Distribución de Edades')
plt.axvline(df['Age'].mean(), color = 'red')
plt.axvline(df['Age'].median(), color = 'green')
plt.axvline(df['Age'].mode()[0], color = 'black')
plt.show()

# Print the results
print(f"Media: {df['Age'].mean()}")
print(f"Mediana: {df['Age'].median()}")
print(f"Moda: {df['Age'].mode()[0]}")
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Age', hue= 'Survived', kde=True, element='step')
plt.title('Distribución de Edad por Superviviencia')
plt.show()
```

```python
plt.figure(figsize=(12, 8))

color = ['#240750', '#344C64']

sns.countplot(data=df, x='Pclass', hue='Survived', palette=color)
plt.title('Supervivencia por clase de pasajero (Pclass)')
plt.xlabel('Clase del pasajero (Pclass)')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Supervivencia', labels=['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

```python
plt.figure(figsize=(12, 8))

color = ['#240750', '#344C64']

sns.countplot(data=df, x='Sex', hue='Survived', palette=color)
plt.title('Supervivencia por sexo (Sex)')
plt.xlabel('Sexo')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Supervivencia', labels=['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

```python
plt.figure(figsize = (12, 8))
sns.heatmap(df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].corr(), 
            annot=True, 
            cbar=True,  # Activar la barra de colores
            cmap='coolwarm',  # Cambiar la paleta de colores
            cbar_kws={'label': 'Nivel de correlación'})  # Etiqueta para la barra de colores
plt.title('Correlación de variables numéricas')
plt.show()
```

## Ejercicios

```python
# Distribución de edades de los pasajeros
plt.figure(figsize=(12, 8))
sns.histplot(data=df, x='Age', kde=True, color='#344C64')
plt.title('Distribución de edades de los pasajeros')
plt.xlabel('Edad (Age)')
plt.ylabel('Cantidad de pasajeros')
plt.show()
```

```python
# Cantidad de sobrevivientes por clase de pasajero (Pclass)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Pclass', hue='Survived', palette=['#240750', '#344C64'])
plt.title('Cantidad de sobrevivientes por clase de pasajero (Pclass)')
plt.xlabel('Clase de pasajero (Pclass)')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Supervivencia', labels=['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

```python
# Cantidad de sobrevivientes por género (Sex)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Sex', hue='Survived', palette=['#240750', '#344C64'])
plt.title('Cantidad de sobrevivientes por género (Sex)')
plt.xlabel('Género (Sex)')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Supervivencia', labels=['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

```python
# Cantidad de sobrevivientes por puerto de embarque (Embarked)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Embarked', hue='Survived', palette=['#240750', '#344C64'])
plt.title('Cantidad de sobrevivientes por puerto de embarque (Embarked)')
plt.xlabel('Puerto de embarque (Embarked)')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Supervivencia', labels=['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

```python
# Categorías de edad y tasa de supervivencia por grupo de edad
bins = [-1, 2, 12, 17, 35, 59, 120]
labels = ['Infante', 'Niño', 'Adolescente', 'Joven Adulto', 'Adulto', 'Mayor']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

tasa_edad = df.groupby('AgeGroup', observed=False)['Survived'].mean() * 100

plt.figure(figsize=(12, 8))
sns.barplot(x=tasa_edad.index, y=tasa_edad.values, color='#344C64')
plt.title('Tasa de supervivencia por grupo de edad (%)')
plt.xlabel('Grupo de edad')
plt.ylabel('Tasa de supervivencia (%)')
plt.ylim(0, 100)
plt.show()
```

```python
# Conteo de pasajeros en cada clase (Pclass)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Pclass', color='#240750')
plt.title('Conteo de pasajeros por clase (Pclass)')
plt.xlabel('Clase de pasajero (Pclass)')
plt.ylabel('Cantidad de pasajeros')
plt.show()
```

```python
# Distribución de género dentro de cada clase de pasajero (Pclass)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Pclass', hue='Sex', palette=['#240750', '#344C64'])
plt.title('Distribución de género dentro de cada clase de pasajero')
plt.xlabel('Clase de pasajero (Pclass)')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Género')
plt.show()
```

```python
# Distribución de clase de pasajero por puerto de embarque (Embarked)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Embarked', hue='Pclass', palette='viridis')
plt.title('Distribución de clase de pasajero por puerto de embarque')
plt.xlabel('Puerto de embarque (Embarked)')
plt.ylabel('Cantidad de pasajeros')
plt.legend(title='Clase (Pclass)')
plt.show()
```

```python
# Conteo de pasajeros por género (Sex)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Sex', color='#344C64')
plt.title('Conteo de pasajeros por género (Sex)')
plt.xlabel('Género (Sex)')
plt.ylabel('Cantidad de pasajeros')
plt.show()
```

```python
# Valores máximo y mínimo de la columna de edad (Age)
edad_min = df['Age'].min()
edad_max = df['Age'].max()
print(f"Edad mínima: {edad_min}")
print(f"Edad máxima: {edad_max}")
```

```python
# Distribución de edades con líneas de media, mediana y moda
media_edad = df['Age'].mean()
mediana_edad = df['Age'].median()
moda_edad = df['Age'].mode()[0]

plt.figure(figsize=(12, 8))
sns.histplot(data=df, x='Age', kde=True, color='#344C64')
plt.axvline(media_edad, color='red', linestyle='--', linewidth=2, label=f'Media: {media_edad:.2f}')
plt.axvline(mediana_edad, color='green', linestyle='-', linewidth=2, label=f'Mediana: {mediana_edad:.2f}')
plt.axvline(moda_edad, color='black', linestyle=':', linewidth=2, label=f'Moda: {moda_edad:.2f}')
plt.title('Distribución de edades (Media, Mediana y Moda)')
plt.xlabel('Edad (Age)')
plt.ylabel('Cantidad de pasajeros')
plt.legend()
plt.show()
```

```python
# Diagrama de caja (boxplot) de la columna de edad (Age)
plt.figure(figsize=(12, 8))
sns.boxplot(data=df, x='Age', color='#344C64')
plt.title('Diagrama de caja de Edad (detección de outliers)')
plt.xlabel('Edad (Age)')
plt.show()
```

```python
# Valores máximo y mínimo de la columna de tarifa (Fare)
tarifa_min = df['Fare'].min()
tarifa_max = df['Fare'].max()
print(f"Tarifa mínima: {tarifa_min}")
print(f"Tarifa máxima: {tarifa_max}")
```

```python
# Distribución de tarifas con líneas de media, mediana y moda
media_fare = df['Fare'].mean()
mediana_fare = df['Fare'].median()
moda_fare = df['Fare'].mode()[0]

plt.figure(figsize=(12, 8))
sns.histplot(data=df, x='Fare', kde=True, color='#344C64')
plt.axvline(media_fare, color='red', linestyle='--', linewidth=2, label=f'Media: {media_fare:.2f}')
plt.axvline(mediana_fare, color='green', linestyle='-', linewidth=2, label=f'Mediana: {mediana_fare:.2f}')
plt.axvline(moda_fare, color='black', linestyle=':', linewidth=2, label=f'Moda: {moda_fare:.2f}')
plt.title('Distribución de tarifas Fare (Media, Mediana y Moda)')
plt.xlabel('Tarifa (Fare)')
plt.ylabel('Cantidad de pasajeros')
plt.legend()
plt.show()
```

```python
# Diagrama de caja (boxplot) de la columna de tarifa (Fare)
plt.figure(figsize=(12, 8))
sns.boxplot(data=df, x='Fare', color='#344C64')
plt.title('Diagrama de caja de Tarifa Fare (detección de outliers)')
plt.xlabel('Tarifa (Fare)')
plt.show()
```

```python
# Filtrar y mostrar pasajeros con tarifa igual a 0
pasajeros_tarifa_cero = df[df['Fare'] == 0]
print(f"Total de pasajeros con tarifa 0: {len(pasajeros_tarifa_cero)}")
display(pasajeros_tarifa_cero)
```

```python
# Conteo de pasajeros por número de hermanos/esposos a bordo (SibSp)
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='SibSp', color='#240750')
plt.title('Conteo de pasajeros por SibSp')
plt.xlabel('Hermanos / Esposos a bordo (SibSp)')
plt.ylabel('Cantidad de pasajeros')
plt.show()
```

```python
# Gráfica de tasa de supervivencia basada en SibSp
tasa_sibsp = df.groupby('SibSp')['Survived'].mean() * 100

plt.figure(figsize=(12, 8))
sns.barplot(x=tasa_sibsp.index, y=tasa_sibsp.values, color='#344C64')
plt.title('Tasa de supervivencia según SibSp (%)')
plt.xlabel('Hermanos / Esposos a bordo (SibSp)')
plt.ylabel('Tasa de supervivencia (%)')
plt.ylim(0, 100)
plt.show()
```

```python
# Porcentaje de supervivencia para cada valor de SibSp
porcentaje_sibsp = (df.groupby('SibSp')['Survived'].mean() * 100).round(2)
print("Porcentaje de supervivencia por valor de SibSp:")
print(porcentaje_sibsp)
```

```python
# Mapa de calor (heatmap) de correlación de características numéricas
plt.figure(figsize=(12, 8))
cols_numericas = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
sns.heatmap(
    df[cols_numericas].corr(),
    annot=True,
    fmt='.2f',
    cbar=True,
    cmap='coolwarm',
    cbar_kws={'label': 'Nivel de correlación'}
)
plt.title('Matriz de correlación de variables numéricas')
plt.show()
```
